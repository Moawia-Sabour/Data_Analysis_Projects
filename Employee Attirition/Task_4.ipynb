# **What if the effect of age on attriration ?**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/Attrition.csv")
```

```python
df
```

```python
#هنحسب متوسط الاعمار لكل فئه
age_stats = df.groupby('Attrition')['Age']

print(age_stats.agg(['count','mean']))
```

```python
#هنرسم رسم يوضح تأثير العمر على البقاء او مغادرة الشركه
plt.figure(figsize=(10, 6))
sns.boxplot(x='Attrition', y='Age', data=df, palette='Set1')

plt.title('Effect of Age on Attrition', fontsize=16, fontweight='bold')
plt.xlabel('Attrition', fontsize=14)
plt.ylabel('Age', fontsize=14)
plt.show()
```

# **Is income the main factor towards employee attrition ?**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/Attrition.csv")
```

```python
plt.figure(figsize=(8, 4))
plt.scatter(df['Attrition'], df['MonthlyIncome'])

plt.xlabel('Attrition')

plt.ylabel('Monthly income')

plt.title('Effect of income to Attrition',fontsize = 16 ,fontweight = 'bold' )

plt.show()
```

```python
income_stats = df.groupby('Attrition')['MonthlyIncome']

print(income_stats.agg(['count','mean']))
```

# **Does the department effect on attrition ?**

```python
df.groupby('Attrition')['Department'].value_counts()
```

```python
ratio_attrition = df.groupby('Department')['Attrition'].value_counts(normalize=True).unstack()
ratio_attrition
```

# **How does Job Satisfaction impact the Attrition ?**

```python
Satisfaction_ratio = df.groupby('JobSatisfaction')['Attrition'].value_counts(normalize=True).unstack()
Satisfaction_ratio
```

```python
Satisfaction_ratio.plot(kind='bar')

plt.title("Attrition Ratio by JobSatisfaction (Stacked)")
plt.xlabel("JobSatisfaction")
plt.ylabel("Satisfaction_ratio")
plt.xticks(rotation=0)

plt.show()
```
